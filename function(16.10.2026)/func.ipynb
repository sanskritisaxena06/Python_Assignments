```python
print("By Sanskriti Saxena")
def show():
    x = 20  # Local Variable
    print("Local Variable: ", x)

show()
print(x)
```

```python
print("By Sanskriti Saxena")
x = 30  # Global Variable

def show():
    x = 20  # Local Variable
    print("Local Variable: ", x)  # 20

show()

print("Global Variable: ", x)  # 30
```

```python
print("By Sanskriti Saxena")
x = 30  # Global Variable

def show():
    global x
    print("Global Variable: ", x)  # 30

show()
print("Global Variable: ", x)  # 30

def display():
    global x
    print("Global Variable: ", x)  # 30

display()
```

```python
print("By Sanskriti Saxena")
balance = 3000  # Global Variable

def show():  # Function without parameter
    deposit = 1000
    print("Deposit Balance: ", deposit)
    print("Initial Balance was: ", globals()["balance"])  # One time use

show()
print("Current Balance After Deposit: ", balance)
```
