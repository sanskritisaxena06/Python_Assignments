```python
# Single Inheritance Example (1 Parent, 1 Child)

class BankAccount():
    def __init__(self, AccountHolderName, balance):
        self.AccountHolderName = AccountHolderName
        self.balance = balance

    def showAccountDetails(self):
        print(self.AccountHolderName)

    def check_Balance(self):
        print("Current Balance: ", self.balance)


class SavingAccount(BankAccount):
    # Constructors, variables, methods

    def add_IntrestRate(self, rate):
        intrest = self.balance * rate / 100
        self.balance += intrest
        print("Interest added: ", intrest)


s1 = SavingAccount("Sanskriti", 3000)
s1.showAccountDetails()
s1.add_IntrestRate(5.0)
s1.check_Balance()

print("by sanskriti saxena")
```

```python
# Multilevel Inheritance (1 Parent, child more than one)

class Employee():
    def __init__(self, Name, Salary):
        self.Name = Name
        self.Salary = Salary

    def ShowEmpDetails(self):
        print("Employee Name: ", self.Name)
        print("Employee Salary: ", self.Salary)


class HR(Employee):
    def calculatePF(self, pf):
        amount = self.Salary - pf
        self.Salary = amount


class Account(Employee):
    def calculatePF(self, pf):
        amount = self.Salary - pf
        self.Salary = amount


h1 = HR("Sanskriti", 30000)
h1.calculatePF(1300)
h1.ShowEmpDetails()

print("=" * 25)

a1 = Account("agrim", 45000)
a1.calculatePF(2400)
a1.ShowEmpDetails()

print("by sanskriti saxena")
```

```python
# Constructor Overriding

class Employee():
    def __init__(self, Name):
        self.Name = Name
        print("Employee Ka Constructor Call Ho Gaya.")


class Developer(Employee):
    def __init__(self, Name, language):
        super().__init__(Name)  # Parent ka Constructor
        self.language = language
        print("Developer Ka Constructor Call Ho Gaya.")

    def showDetails(self):
        print("Employee Name: ", self.Name)
        print("Language: ", self.language)


d1 = Developer("Sanskriti", "Python")  # __init__() Child
d1.showDetails()


# issubclass()
# isinstance()

print(isinstance(d1, Developer))
print(issubclass(Developer, Employee))
print(issubclass(Employee, Developer))

print("by sanskriti saxena")
```
