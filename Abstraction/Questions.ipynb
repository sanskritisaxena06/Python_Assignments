```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):  # Method Overriding
        print("Dog Barks")

class Cat(Animal):
    def sound(self):  # Method Overriding
        print("Cat Meows")

d1 = Dog()
d1.sound()

c1 = Cat()
c1.sound()

print("by sanskriti saxena")
```
