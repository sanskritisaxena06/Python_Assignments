```python
# Encapsulation
# data private
# functions public

# Access Modifiers
# public
# private
# protected

print("by sanskriti saxena")
```

```python
class Employee():
    def addDetails(self):
        self.balance = 10000  # public


e1 = Employee()
e1.addDetails()
print(e1.balance)

print("by sanskriti saxena")
```

```python
class Employee():
    def addDetails(self):
        self._balance = 10000  # protected


e1 = Employee()
e1.addDetails()
print(e1._balance)

print("by sanskriti saxena")
```

```python
class Employee():
    def addDetails(self):
        self.__balance = 10000  # private


e1 = Employee()
e1.addDetails()

try:
    print(e1.__balance)
except AttributeError as e:
    print(e)

print("by sanskriti saxena")
```

```python
class Employee():
    def addDetails(self, Name, Profile, Salary):
        self.__Name = Name
        self.__Profile = Profile
        self.__Salary = Salary

    def showEmpDetails(self):
        print("Employee Name: ", self.__Name)
        print("Employee Profile: ", self.__Profile)
        print("Employee Salary: ", self.__Salary)


e1 = Employee()
e1.addDetails("Sanskriti", "Data Scientist", 20000)
e1.showEmpDetails()

print("by sanskriti saxena")
```

```python
class Employee():
    def __init__(self, Name, Profile, Salary):
        self.__Name = Name
        self.__Profile = Profile
        self.__Salary = Salary

    def setName(self, Name):
        self.__Name = Name

    def setProfile(self, Profile):
        self.__Profile = Profile

    def setSalary(self, Salary):
        self.__Salary = Salary

    def getName(self):
        return self.__Name

    def getProfile(self):
        return self.__Profile

    def getSalary(self):
        return self.__Salary


e1 = Employee("Sanskriti", "HR", 20000)

e1.setName("Sanskriti Saxena")  # Setter used to Update Object Name
e1.setSalary(27000)

print(e1.getName())
print(e1.getProfile())
print(e1.getSalary())

print("by sanskriti saxena")
```
