```python
print("By Sanskriti Saxena")
def CalaculateBill(price, discount=0):
    discount_amount = price * discount / 100
    bill = price - discount_amount

    print("Original Price: ", price)
    print("Discount : ", discount, "%")
    print("Discount Amount: ", discount_amount)
    print("Final Bill Amount: ", bill)
    print("============================")

CalaculateBill(20000)  # Cash 0% discount
CalaculateBill(10000, 10)  # UPI 10% discount
CalaculateBill(30000, 20)  # Card 20% discount
```

```python
print("By Sanskriti Saxena")
def Employeedetails(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

    print("======================")

Employeedetails(name="Sanskriti", Role="Developer", Department="IT")
Employeedetails(name="Sanskriti", Department="IT")
Employeedetails(name="Sanskriti")
```

```python
# args
print("By Sanskriti Saxena")
def add(*args):  # tuple
    total = 0

    for ex in args:
        total = total + ex

    print("Sum: ", total)

add(10, 20)  # args = (10, 20)
add(40, 50, 60)  # args = (40, 50, 60)
add(20, 30, 40, 50)  # args = (20, 30, 40, 50)
add(20, 30, 40, 50, 60)  # args = (20, 30, 40, 50, 60)
```

```python
print("By Sanskriti Saxena")
def add(a, b):
    return a + b

def add(a, b, c):
    return a + b + c

def add(a, b, c, d):
    return a + b + c + d

def add(a, b, c, d, e):
    return a + b + c + d + e

add(10, 20)  # 2
add(20, 30, 40)  # 3
add(20, 30, 40, 50)  # 4
add(20, 30, 40, 50, 60)  # 5
```

```python
print("By Sanskriti Saxena")
def add(*args):
    return sum(args)

print(add(10, 20))                 # 2 arguments
print(add(20, 30, 40))             # 3 arguments
print(add(20, 30, 40, 50))         # 4 arguments
print(add(20, 30, 40, 50, 60))     # 5 arguments
```
