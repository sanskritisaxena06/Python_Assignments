```python
print("By Sanskriti Saxena")

# Real-Time Project 2: Bank Account Management
#
# Requirements:
# 1. Set the account holder's name and opening balance.
# 2. Deposit money into the account.
# 3. Withdraw money from the account.
# 4. Check the current account balance.
# 5. Maintain each account's balance independently.


class BankAccount:

    # Constructor to initialize account details
    def __init__(self, account_number, account_holder_name, balance=0):
        self.account_number = account_number
        self.account_holder_name = account_holder_name
        self.balance = balance

    # Deposit money into account
    def deposit(self, amount):
        if amount > 0:
            self.balance = self.balance + amount
            print("Amount Deposited:", amount)
            print("Deposit Successful")
        else:
            print("Invalid deposit amount")

    # Withdraw money from account
    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
        elif amount > self.balance:
            print("Insufficient Balance")
        else:
            self.balance = self.balance - amount
            print("Amount Withdrawn:", amount)
            print("Withdrawal Successful")

    # Check current balance
    def check_balance(self):
        print("Current Balance:", self.balance)

    # Display account details
    def show_account_details(self):
        print("Account Number:", self.account_number)
        print("Account Holder Name:", self.account_holder_name)
        print("Current Balance:", self.balance)


# Creating first bank account
a1 = BankAccount(123123, "Sanskriti", 5000)

a1.show_account_details()
print("=" * 40)

# Deposit money
a1.deposit(2000)
a1.check_balance()

print("=" * 40)

# Withdraw money
a1.withdraw(1500)
a1.check_balance()

print("=" * 40)

# Creating second bank account
a2 = BankAccount(4633252, "Anu", 2000)

a2.show_account_details()
print("=" * 40)

# Deposit money into second account
a2.deposit(1000)
a2.check_balance()

print("=" * 40)

# Withdraw money from second account
a2.withdraw(500)
a2.check_balance()

print("=" * 40)
```

```python
print("By Sanskriti Saxena")

# Real-Time Project 3: Employee Payroll System
#
# Requirements:
# 1. Store the employee ID, name, and basic salary.
# 2. Calculate HRA = 20% of the basic salary.
# 3. Calculate DA = 10% of the basic salary.
# 4. Calculate PF deduction = 12% of the basic salary.
# 5. Calculate net salary = Basic Salary + HRA + DA - PF.
#
# Note:
# These percentages are used for demonstration purposes.


class Employee:

    # Constructor to initialize employee details
    def __init__(self, employee_id, name, basic_salary):
        self.employee_id = employee_id
        self.name = name
        self.basic_salary = basic_salary

    # Calculate HRA = 20% of basic salary
    def calculate_hra(self):
        return self.basic_salary * 20 / 100

    # Calculate DA = 10% of basic salary
    def calculate_da(self):
        return self.basic_salary * 10 / 100

    # Calculate PF = 12% of basic salary
    def calculate_pf(self):
        return self.basic_salary * 12 / 100

    # Calculate net salary
    def calculate_net_salary(self):
        hra = self.calculate_hra()
        da = self.calculate_da()
        pf = self.calculate_pf()

        return self.basic_salary + hra + da - pf

    # Display complete employee payroll
    def show_payroll(self):
        print("Employee ID:", self.employee_id)
        print("Employee Name:", self.name)
        print("Basic Salary:", self.basic_salary)
        print("HRA (20%):", self.calculate_hra())
        print("DA (10%):", self.calculate_da())
        print("PF (12%):", self.calculate_pf())
        print("Net Salary:", self.calculate_net_salary())


# Creating first employee
e1 = Employee(111, "Anu", 12000)

# Display first employee payroll
e1.show_payroll()

print("=" * 40)

# Creating second employee
e2 = Employee(222, "Sanskriti", 22000)

# Display second employee payroll
e2.show_payroll()

print("=" * 40)
```
