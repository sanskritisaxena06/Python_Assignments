```python
# Question 6: Create a Student Management System.
# Requirements:
# 1. Create student objects.
# 2. Initialize student details using a constructor.
# 3. Store student marks.
# 4. Calculate total marks and average marks.
# 5. Calculate the student's grade.
# 6. Display the complete student report.

print("by sanskriti saxena")

class Student:
    def __init__(self, roll_no, name, marks):
        self.roll_no = roll_no
        self.name = name
        self.marks = marks

    def calculate_total(self):
        return sum(self.marks)

    def calculate_average(self):
        return self.calculate_total() / len(self.marks)

    def calculate_grade(self):
        average = self.calculate_average()

        if average >= 90:
            return "A+"
        elif average >= 80:
            return "A"
        elif average >= 70:
            return "B"
        elif average >= 60:
            return "C"
        elif average >= 40:
            return "D"
        else:
            return "F"

    def display_report(self):
        print("Student Roll Number:", self.roll_no)
        print("Student Name:", self.name)
        print("Marks:", self.marks)
        print("Total Marks:", self.calculate_total())
        print("Average Marks:", round(self.calculate_average(), 2))
        print("Grade:", self.calculate_grade())


s1 = Student(101, "Sanskriti", [85, 90, 88, 92, 87])
s1.display_report()

print("=" * 30)

s2 = Student(102, "Agrim", [75, 80, 78, 82, 70])
s2.display_report()
```

```python
# Question 7: Create a Bank Account Management System.
# Requirements:
# 1. Set the account holder's name and opening balance.
# 2. Deposit money into the account.
# 3. Withdraw money from the account.
# 4. Check the current account balance.
# 5. Maintain each account's balance independently.

print("by sanskriti saxena")

class BankAccount:
    def __init__(self, account_number, account_holder_name, balance=0):
        self.account_number = account_number
        self.account_holder_name = account_holder_name
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print("Amount deposited successfully:", amount)
        else:
            print("Please enter a valid deposit amount.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Please enter a valid withdrawal amount.")
        elif amount > self.balance:
            print("Insufficient balance.")
        else:
            self.balance -= amount
            print("Amount withdrawn successfully:", amount)

    def check_balance(self):
        print("Current Balance:", self.balance)

    def show_account_details(self):
        print("Account Number:", self.account_number)
        print("Account Holder Name:", self.account_holder_name)
        print("Current Balance:", self.balance)


account1 = BankAccount(123123, "Sanskriti", 5000)
account1.show_account_details()

account1.deposit(2000)
account1.withdraw(1500)
account1.check_balance()

print("=" * 30)

account2 = BankAccount(4633252, "Agrim", 2000)
account2.show_account_details()

account2.deposit(1000)
account2.withdraw(500)
account2.check_balance()
```

```python
# Question 8: Create an Employee Payroll System.
# Requirements:
# 1. Store employee ID, name, and basic salary.
# 2. Calculate HRA = 20% of basic salary.
# 3. Calculate DA = 10% of basic salary.
# 4. Calculate PF deduction = 12% of basic salary.
# 5. Calculate net salary = Basic Salary + HRA + DA - PF.
# Note: These percentages are for demonstration purposes.

print("by sanskriti saxena")

class EmployeePayroll:
    def __init__(self, employee_id, name, basic_salary):
        self.employee_id = employee_id
        self.name = name
        self.basic_salary = basic_salary

    def calculate_hra(self):
        return self.basic_salary * 0.20

    def calculate_da(self):
        return self.basic_salary * 0.10

    def calculate_pf(self):
        return self.basic_salary * 0.12

    def calculate_net_salary(self):
        return (
            self.basic_salary
            + self.calculate_hra()
            + self.calculate_da()
            - self.calculate_pf()
        )

    def display_salary_slip(self):
        print("Employee ID:", self.employee_id)
        print("Employee Name:", self.name)
        print("Basic Salary:", round(self.basic_salary, 2))
        print("HRA (20%):", round(self.calculate_hra(), 2))
        print("DA (10%):", round(self.calculate_da(), 2))
        print("PF Deduction (12%):", round(self.calculate_pf(), 2))
        print("Net Salary:", round(self.calculate_net_salary(), 2))


e1 = EmployeePayroll(111, "Sanskriti", 20000)
e1.display_salary_slip()

print("=" * 30)

e2 = EmployeePayroll(222, "Sanskriti Saxena", 30000)
e2.display_salary_slip()
```
